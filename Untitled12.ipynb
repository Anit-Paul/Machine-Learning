```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
tokenizer.word_index
```

```python
for sentence in data.split('\n'):
    print(tokenizer.texts_to_sequences([sentence])[0])
```

```python
#forming the data set
x_train=[]
y_train=[]
for sentence in data.split('\n'):
    tokenized_sentence=tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1,len(tokenized_sentence)):
        x_train.append(tokenized_sentence[:i])
        y_train.append(tokenized_sentence[i])
```

```python
max_len=max([len(x) for x in x_train])
```

```python
max_len
```

```python
#for making all the rows of equal size we need to add zero padding
x_train=pad_sequences(x_train,maxlen=max_len,padding='pre')
```

```python
from tensorflow.keras.utils import to_categorical
y_train=to_categorical(y_train,num_classes=len(tokenizer.word_index)+1)
```

```python
y_train.shape
```

```python
model=Sequential()
model.add(Embedding(
    input_dim=len(tokenizer.word_index) + 1,
    output_dim=100,
    input_shape=(max_len,)
))
model.add(LSTM(150))
model.add(Dense(len(tokenizer.word_index)+1,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=100)
```

```python

text="Vijay is introduced to"
#tokenizer
tokenized_text=tokenizer.texts_to_sequences([text])[0]
#padding
padded_text=pad_sequences([tokenized_text],maxlen=max_len,padding='pre')
#prediction
model.predict(padded_text)
```

```python
import numpy as np
def make_prediction(text):
  #tokenizer
  tokenized_text=tokenizer.texts_to_sequences([text])[0]
  #padding
  padded_text=pad_sequences([tokenized_text],maxlen=max_len,padding='pre')
  idx=np.argmax(model.predict(padded_text))
  for i,j in tokenizer.word_index.items():
    if j==idx:
      print(i)
      break
```
