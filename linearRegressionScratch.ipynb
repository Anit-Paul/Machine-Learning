```python
import pandas as pd 
import numpy as np 
```

```python
from sklearn.datasets import load_diabetes

data = load_diabetes()

x = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split

# X represents your features, y represents your target labels
x_train, x_test, y_train, y_test = train_test_split(
    x, y, 
    test_size=0.2,       # Allocates 20% of data for testing
    random_state=42,     #Ensures reproducibility so the split is identical every run
    shuffle=True         # Randomizes the data before splitting
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#regression model 
#beta=(xtx)-1 * xty
def get_coeff(x_train,y_train):
    xt=x_train.transpose()
    xtx=np.dot(xt,x_train)
    XtX_inv = np.linalg.inv(xtx)
    return np.dot(XtX_inv,np.dot(xt,y_train))
```

```python
coeff=get_coeff(x_train,y_train)
```

```python
coeff
```

```python
y_pred=np.dot(x_test,coeff)
```

```python
#mean square error 
y_pred.shape
```

```python
n=y_pred.shape[0]
s=0
for i in range(n):
    s+=((y_test[i]-y_pred[i])**2)
print(s/n)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
from sklearn.metrics import mean_squared_error
```
