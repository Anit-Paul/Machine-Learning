```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv('Churn_Modelling.csv')
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.head()
```

```python
df.drop(['RowNumber','CustomerId','Surname'],axis=1,inplace=True)
```

```python
df.head()
```

```python
#one hot encoding 
df=pd.get_dummies(df,['Geography','Gender'],drop_first=True)
```

```python
df
```

```python
x=df.drop('Exited',axis=1)
y=df['Exited']
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=2
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
#!pip install tensorflow
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(10,activation='relu',input_dim=11)) #for hidden 
model.add(Dense(10,activation='relu'))
model.add(Dense(1,activation='sigmoid')) # for output
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='Adam')
```

```python
history=model.fit(x_train_scaled,y_train,validation_data=(x_test,y_test),epochs=500)
```

```python
plt.plot(history.history["loss"],label="train")
plt.plot(history.history["val_loss"],label="test")
plt.legend()
plt.show()
```

```python
model.layers[0].get_weights()
```

```python
y_log=model.predict(x_test_scaled)
```

```python
y_pred=np.where(y_log>0.5,1,0)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
